```python
import flap
```

```python
defl0 = flap.load('../pickled_shot_data/20250409.046_raw_defl0.pkl')
defl1 = flap.load('../pickled_shot_data/20250409.046_raw_defl1.pkl')
```

```python
f = 2000

defl0_f = defl0.filter_data(
    coordinate = 'Time',
    options = {
        'Design': 'Butterworth',
        'Type': 'Highpass',
        'f_low': f,
        'Steepness': 0.5,
        'Loss': 1,
        'Attenuation': 40
    }
)

defl1_f = defl1.filter_data(
    coordinate = 'Time',
    options = {
        'Design': 'Butterworth',
        'Type': 'Highpass',
        'f_low': f,
        'Steepness': 0.5,
        'Loss': 1,
        'Attenuation': 40
    }
)
```

```python
flap.save(defl0_f, f'../pickled_shot_data/20250409.046_highpass_{f}_defl0.pkl')
flap.save(defl1_f, f'../pickled_shot_data/20250409.046_highpass_{f}_defl1.pkl')
```
