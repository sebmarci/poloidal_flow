```python
import flap
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

mpl.rcParams.update({
    'font.size': 12
})
```

```python
defl0 = flap.load('../pickled_shot_data/20250409.046_raw_defl0.pkl')
defl1 = flap.load('../pickled_shot_data/20250409.046_raw_defl1.pkl')

defl0 = defl0.slice_data(
    slicing = {'Time': flap.Intervals(6, 7)}
)

defl1 = defl1.slice_data(
    slicing = {'Time': flap.Intervals(6, 7)}
)
```

```python
defl0_timeslice = defl0.slice_data(
    slicing = {'Time': 6.6}
)
```

```python
times0 = defl0.coordinate('Time')[0][0]
times1 = defl1.coordinate('Time')[0][0]
chs = np.arange(1, 41)
```

```python
fig, (ax1, ax2) = plt.subplots(nrows = 1, ncols = 2, figsize = (15, 6))

ax1.set_xlabel('Channel')
ax1.set_ylabel('Time [s]')

ax2.set_xlabel('Channel')
ax2.set_ylabel('Light intensity')

pcm = ax1.pcolormesh(
    chs,
    times0,
    defl0.data.T,
    shading = 'nearest',
    cmap = 'plasma'
)

ax2.plot(
    defl0_timeslice.coordinate('Channel number')[0],
    defl0_timeslice.data
)

ax1.set_title('Full profile, t = 6-7 s')
ax2.set_title('Time slice, t = 6.6 s')

ax1.axhline(6.6, c = 'g', ls = '--', lw = 3)

cbar = fig.colorbar(pcm, ax = ax1)
cbar.set_label('Light intensity')

plt.suptitle('20250409.046 ABES light profile, defl = 0')
```
